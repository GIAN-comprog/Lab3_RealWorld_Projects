```python
# [CELL 1] GLOBAL STUDENT CONFIGURATION
LAST_NAME = "BERNARDO"
STUDENT_ID = "TUPM-26-4349"
FAVORITE_ARTIST = "RADIOHEAD"

# Extracted variables using the exact conventions from your screens
SEED_DIGIT = int(STUDENT_ID[-1]) # Evaluates to 9
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # Evaluates to 28
NAME_LENGTH = len(LAST_NAME) # Evaluates to 8
ARTIST_LENGTH = len(FAVORITE_ARTIST) # Evaluates to 9

print("--- INITIALIZATION COMPLETED ---")
print(f"LAST_NAME: {LAST_NAME} (Length: {NAME_LENGTH})")
print(f"SEED_DIGIT: {SEED_DIGIT} | ID_SUM: {ID_SUM}")
print(f"FAVORITE_ARTIST: {FAVORITE_ARTIST} (Length: {ARTIST_LENGTH})")
```

```python
# [CELL 2] EXERCISE 1: DECORATOR DEFINITION
def diagnostic_logger(func):
    """Decorator to record the diagnostic process as required."""
    def wrapper(*args, **kwargs):
        print("\n--- [LOG START] Initiating Diagnostic System Check ---")
        result = func(*args, **kwargs)
        print("--- [LOG END] Process safely wrapped down ---")
        return result
    return wrapper
```

```python
# [CELL 3] EXERCISE 1: REUSABLE LOGIC CORE
def validate_reading(val):
    """Validates if the structural entry is a normal integer."""
    if not isinstance(val, (int, float)):
        raise TypeError("Invalid entry datatype detected!")
    if val < 0:
        raise ValueError("Out of bounds metric: negative reading!")
    return True

def calculate_metric(val):
    """Applies computational steps adjusted by your student configurations."""
    return round((val * 1.05) + (SEED_DIGIT / 2), 2)

def classify_status(metric):
    """Classifies system severity tiers."""
    threshold = NAME_LENGTH * 10
    if metric > threshold:
        return "CRITICAL FAULT DETECTED"
    return "OPERATIONAL / STABLE"
```

```python
# [CELL 4] EXERCISE 1: EXECUTION ENVIRONMENT
@diagnostic_logger
def run_equipment_diagnostic():
    # Student specific equipment readings dataset generation
    raw_readings = [10, 45, "MALFUNCTION_STRING", 85, -5, 120]
    
    valid_readings = []
    calc_results = []
    classifications = []
    execution_errors = 0
    
    print(f"Processing unique dataset for student: {LAST_NAME.upper()}")
    
    for entry in raw_readings:
        try:
            validate_reading(entry)
            
            # Processing steps if valid
            computed = calculate_metric(entry)
            status = classify_status(computed)
            
            valid_readings.append(entry)
            calc_results.append(computed)
            classifications.append(status)
            
        except (TypeError, ValueError) as error:
            execution_errors += 1
            # Continues processing remaining readings loop without program termination
            print(f"[EXCEPT WARNING] Skipped item '{entry}': {error}")
            continue

    # Summary Generation Output Section
    print("\n====== Assessment Data Summary ======")
    print("Generated Equipment Data:", raw_readings)
    print("Validation Results (Passed):", valid_readings)
    print("Diagnostic Results:", classifications)
    print("Execution Log Errors Blocked:", execution_errors)
    print(f"Final Output Summary: {len(valid_readings)} systems resolved successfully.")

run_equipment_diagnostic()
```

```python
# [CELL 5] EXERCISE 2: INITIALIZATION AND COUNT TRACKER
# Generate a student-specific numeric fault code
FAULT_CODE = (NAME_LENGTH * SEED_DIGIT) + ARTIST_LENGTH # (8 * 9) + 9 = 81
recursive_calls_counter = 0
```

```python
# [CELL 6] EXERCISE 2: RECURSIVE ARCHITECTURE
def process_fault_trace(current_code):
    global recursive_calls_counter
    recursive_calls_counter += 1
    
    # 3. Define and apply an appropriate base condition
    if current_code <= 0:
        print(f"Base Condition Verified! Level reached: {current_code}")
        return [current_code]
    
    # Custom formula degradation stepping down based on your seed variables
    step_down = SEED_DIGIT * 2 # Step size = 18
    next_code = current_code - step_down
    
    # Recursive trace pattern building step
    return [current_code] + process_fault_trace(next_code)
```

```python
# [CELL 7] EXERCISE 2: RUN SYSTEM TRACE
trace_output_log = process_fault_trace(FAULT_CODE)

print("\n====== Assessment Data Summary ======")
print("Generated Fault Data Target:", FAULT_CODE)
print("Recursive Trace History Map:", trace_output_log)
print("Number of Recursive Calls:", recursive_calls_counter)
print("Execution Log: Recursive processing executed cleanly through complete lifecycle.")
print("Final Output Result: Final safe state confirmed at index value:", trace_output_log[-1])
```

```python
# [CELL 9] EXCERCISE 3 MODULE B: MONITOR DECORATOR & LAMBDA FILTER
def pipeline_decorator_monitor(func):
    """Monitors the execution context of the main data loop."""
    def inner_run(*args, **kwargs):
        print("\n>>> PIPELINE ACTIVATION METRIC CHECK UPON MONITOR UNIT <<<")
        func_return = func(*args, **kwargs)
        print(">>> PIPELINE PROCESSING TERMINATED INDEPENDENTLY <<<\n")
        return func_return
    return inner_run

# Lambda expression to filter or transform data thresholds
is_abnormal_reading = lambda value: value > 50.0
```

```python
# [CELL 8] EXCERCISE 3 MODULE A: DATA PIPELINE STREAM GENERATOR
def live_telemetry_generator(limit_range):
    """Produces telemetry values instantly without storing whole arrays in memory."""
    for step in range(1, limit_range + 1):
        # Programmatic stream generation derived by student variables
        simulated_reading = (step * 8) + (ID_SUM % 5)
        
        # Inject artificial issues for exceptions validation testing
        if step == 5:
            yield -999.0 # Poison pill invalid reading marker
        else:
            yield float(simulated_reading)
```

```python
# [CELL 10] MAIN INTEGRATED EXECUTION LOGIC
@pipeline_decorator_monitor
def execute_monitoring_pipeline():
    total_processed = 0
    valid_readings_count = 0
    invalid_readings_count = 0
    abnormalities_detected = 0
    
    # Variable operational constraint parameters
    stream_size = ARTIST_LENGTH + 5 # 9 + 5 = 14 samples
    
    print(f"Pipeline Configured for Target Domain Profile: {LAST_NAME.upper()}")
    
    # Access generator loop stream directly
    for tele_value in live_telemetry_generator(stream_size):
        total_processed += 1
        
        try:
            # Exception Handlings: Catch unexpected metrics instantly
            if tele_value < 0:
                raise ValueError(f"CRITICAL SYSTEM BREAKDOWN INTERRUPT: Negative bounds '{tele_value}'")
            
            valid_readings_count += 1
            
            # Use logic filters
            if is_abnormal_reading(tele_value):
                abnormalities_detected += 1
                
        except ValueError as exception_message:
            invalid_readings_count += 1
            # Keeps the processing continuous without dropping the whole telemetry module runtime
            print(f"[Pipeline Exception Logged] -> {exception_message}")
            continue

    # Final overall condition balance assessment logic
    if abnormalities_detected > 3:
        overall_status = "CRITICAL: DISPATCH MAINTENANCE IMMEDIATE"
    else:
        overall_status = "STABLE OPERATIONAL PARAMETERS"
        
    print("\n====== Final Diagnostic Report Output ======")
    print(f"Number of Processed Readings: {total_processed}")
    print(f"Valid Readings Recorded : {valid_readings_count}")
    print(f"Invalid Readings Dropped : {invalid_readings_count}")
    print(f"Detected Abnormal Conditions: {abnormalities_detected}")
    print(f"Overall Equipment Status : {overall_status}")

# Trigger full integration run
execute_monitoring_pipeline()
```
